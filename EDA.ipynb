# **NumPy**

```python
import numpy as np
```

**np.array() → Membuat array dari daftar atau tuple**

Digunakan untuk mengubah data menjadi array NumPy agar lebih mudah diolah.

```python
data = [10, 20, 30, 40, 50]
array_data = np.array(data)
print(array_data)
```

**np.mean() → Menghitung rata-rata**

Digunakan untuk mengetahui nilai rata-rata dan nilai tengah dari dataset.

```python
data = np.array([12, 15, 14, 10, 18])
mean_value = np.mean(data)
median_value = np.median(data)
print(mean_value)
print(median_value)
```

**np.std() dan np.var()**

np.std() = Berguna untuk menghitung standar deviasi

np.var() = Berguna untuk menghitung varians

np.std() berfungsi untuk melihat sebaran data terhadap rata-rata.

```python
data = np.array([5, 10, 15, 20, 25])
std_value = np.std(data)
var_value = np.var(data)
print(std_value)
print(var_value)
```

**np.min() & np.max()**

Menentukan nilai minimum dan maksimum

```python
data = np.array([4, 7, 1, 9, 12])
min_value = np.min(data)
max_value = np.max(data)

print(f"Min: {min_value}, Max: {max_value}")
```

**np.corrcoef()**

Untuk menghitung korelasi antar dua dataset

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

correlation = np.corrcoef(x, y)
print(correlation)
```

**np.random.randn()**

Untuk membuat data acak dengan distribusi normal

```python
random_data = np.random.randn(5)
print(random_data)
```

Dalam EDA, NumPy digunakan untuk berbagai analisis statistik, termasuk rata-rata, standar deviasi, varians, korelasi, dan histogram. Fungsi-fungsi ini membantu dalam memahami distribusi dan pola data sebelum membuat model lebih lanjut.

## Categorical Data vs Churn

```python
def proportional_aggregation(df, y, x):
    temp_df = df.groupby([y,x], as_index = False).size()
    temp_df['proportional'] = temp_df['size'] / temp_df.groupby(y)['size'].transform('sum')
    return temp_df
```

```python
temp_raw = raw_data.groupby(['Churn','gender'], as_index = False).size()
temp_raw
```

```python
proportional_aggregation(raw_data, "Churn", "gender")
```

```python
import matplotlib.pyplot as plt

data_proportional = proportional_aggregation(raw_data, "Churn", "gender")

plt.figure(figsize=(15, 7))
sns.barplot(x="gender", y="proportional", hue="Churn", data=data_proportional)
plt.title("Proportional Churn by Gender")
plt.ylabel("Proportional")
plt.xlabel("Gender")
plt.show()
```

**Senior Citizen vs Churn**

```python
g = sns.FacetGrid(
    data = proportional_aggregation(raw_data, "SeniorCitizen", "Churn"), #pilih kolom yang ingin dihitung dan disimpan didalam variabel data
    col = "SeniorCitizen", #Pilih kolom yang ingin dianalisa
    aspect = 2
    #Nilai 2 mengindikasikan bahwa subplot akan memiliki lebar yang dua kali lebih besar dari tingginya.
)
g.map(sns.barplot, "Churn", "proportional");
#mengisi data kedalam grid
```
